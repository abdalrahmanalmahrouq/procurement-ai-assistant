```python
orders_v2 = (
    df.groupby(
        [
            "Department Name",
            "Purchase Order Number"
        ],
        dropna=False
    )
    .agg(
        row_count=("Purchase Order Number", "size"),
        creation_dates=("Creation Date", "nunique"),
        suppliers=("Supplier Code", "nunique"),
        total_order_value=(
            "Total Price",
            lambda x: x.sum(min_count=1)
        )
    )
    .reset_index()
)
```

```python
orders_lpa = (
    df.groupby(
        [
            "Department Name",
            "Purchase Order Number",
            "Creation Date"
        ],
        dropna=False
    )
    .agg(
        total_order_value=(
            "Total Price",
            lambda x: x.sum(min_count=1)
        ),
        has_lpa=(
            "LPA Number",
            lambda x: (
                x.fillna("")
                 .astype(str)
                 .str.strip()
                 .ne("")
                 .any()
            )
        )
    )
    .reset_index()
)
```

```python
orders_lpa["below_5000"] = (
    orders_lpa["total_order_value"] < 5000
)
```

```python
orders_lpa.groupby("has_lpa")["below_5000"].agg(
    ["count", "sum", "mean"]
)
```

```python
numeric_columns = [
    "Quantity",
    "Unit Price",
    "Total Price"
]

for column in numeric_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
df.info()
```

```python
df[numeric_columns].isna().sum()
```

```python
df[numeric_columns].describe()
```

```python
df[
    [
        "Quantity",
        "Unit Price",
        "Total Price"
    ]
].head(20)
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_", regex=False)
      .str.replace("-", "_", regex=False)
)
```

```python
df.columns.tolist()
```

```python
df["creation_date"] = pd.to_datetime(
    df["creation_date"],
    errors="coerce"
)

df["purchase_date"] = pd.to_datetime(
    df["purchase_date"],
    errors="coerce"
)
```

```python
df["order_key"] = (
    df["department_name"].fillna("UNKNOWN").astype(str)
    + "|"
    + df["purchase_order_number"].fillna("UNKNOWN").astype(str)
    + "|"
    + df["creation_date"]
        .dt.strftime("%Y-%m-%d")
        .fillna("UNKNOWN")
)
```

```python
df["year"] = df["creation_date"].dt.year
df["month"] = df["creation_date"].dt.month
df["quarter"] = df["creation_date"].dt.quarter
```
